```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
categories = [
    "rec.sport.baseball",
    "rec.sport.hockey",
    "sci.space",
    "comp.graphics"
]
train_data = fetch_20newsgroups(
    subset="train",
    categories=categories,
    remove=("headers", "footers", "quotes")
)
test_data = fetch_20newsgroups(
    subset="test",
    categories=categories,
    remove=("headers", "footers", "quotes")
)
print("Number of Training Documents :", len(train_data.data))
print("Number of Testing Documents  :", len(test_data.data))
print("\nClasses:")
for i, category in enumerate(train_data.target_names):
    print(i, "->", category)
model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            stop_words="english",
            max_features=10000
        )
    ),
    (
        "classifier",
        MultinomialNB()
    )
])
model.fit(
    train_data.data,
    train_data.target
)
print("\nModel Training Completed Successfully!")
predicted = model.predict(test_data.data)
accuracy = accuracy_score(
    test_data.target,
    predicted
)
print("\nAccuracy :", accuracy)

print("\nAccuracy Percentage :",
      round(accuracy * 100, 2), "%")
print("\nClassification Report")
print("==============================")
print(
    classification_report(
        test_data.target,
        predicted,
        target_names=test_data.target_names
    )
)
print("\nConfusion Matrix")
print("==============================")
cm = confusion_matrix(
    test_data.target,
    predicted
)
print(cm)
sample_documents = [
    "The baseball team won the match after scoring five runs.",
    "The hockey player scored a goal in the final period.",
    "NASA launched a spacecraft to study the planets.",
    "The computer graphics system creates realistic images."
]
sample_predictions = model.predict(sample_documents)
print("\nSample Document Classification")
print("==============================")
for document, prediction in zip(
    sample_documents,
    sample_predictions
):

    print("\nDocument :", document)
    print("Predicted Class :",
          train_data.target_names[prediction])
print("\nExperiment Completed Successfully!")
```

```python
# Naive Bayes Text Classification

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

categories = [
    "rec.sport.baseball",
    "rec.sport.hockey",
    "sci.space",
    "comp.graphics"
]

train = fetch_20newsgroups(
    subset="train", categories=categories,
    remove=("headers", "footers", "quotes")
)

test = fetch_20newsgroups(
    subset="test", categories=categories,
    remove=("headers", "footers", "quotes")
)

model = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),
    ("nb", MultinomialNB())
])

model.fit(train.data, train.target)

predicted = model.predict(test.data)

print("Naive Bayes Classification")
print("Training Documents :", len(train.data))
print("Testing Documents  :", len(test.data))
print("Accuracy :", round(accuracy_score(
    test.target, predicted) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    test.target,
    predicted,
    target_names=test.target_names
))
```

```python
# SVM Text Classification

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

categories = [
    "rec.sport.baseball",
    "rec.sport.hockey",
    "sci.space",
    "comp.graphics"
]

train = fetch_20newsgroups(
    subset="train", categories=categories,
    remove=("headers", "footers", "quotes")
)

test = fetch_20newsgroups(
    subset="test", categories=categories,
    remove=("headers", "footers", "quotes")
)

model = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),
    ("svm", LinearSVC())
])

model.fit(train.data, train.target)

predicted = model.predict(test.data)

print("SVM Classification")
print("Training Documents :", len(train.data))
print("Testing Documents  :", len(test.data))
print("Accuracy :", round(accuracy_score(
    test.target, predicted) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    test.target,
    predicted,
    target_names=test.target_names
))
```

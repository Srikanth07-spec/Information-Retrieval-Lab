```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Documents
documents = [
    "information retrieval is used to search information",
    "information retrieval is used to find information",
    "machine learning is used for data analysis"
]

# Create vocabulary
vectorizer = CountVectorizer(binary=True)
matrix = vectorizer.fit_transform(documents)

# Vocabulary
vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(list(vocabulary))

# Binary Document Vectors
print("\nDocument Vectors:")
for i, vector in enumerate(matrix.toarray(), 1):
    print("Document", i, ":", vector.tolist())

# Cosine Similarity
similarity_matrix = cosine_similarity(matrix)

print("\nSimilarity between Document 1 and Document 2:",
      round(float(similarity_matrix[0][1]), 4))

# Similarity Matrix
print("\nSimilarity Matrix:")
for row in similarity_matrix:
    print([round(float(x), 4) for x in row])
```

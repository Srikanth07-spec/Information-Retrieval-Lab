```python
import os
import re
from collections import defaultdict

# Folder containing the text documents
folder = "documents"

# Common stop words
stop_words = {
    "the", "is", "a", "an", "and", "or", "of", "to", "in", "on",
    "for", "with", "by", "as", "at", "from", "that", "this",
    "it", "are", "was", "be", "has", "have"
}

# Dictionary to store the inverted index
inverted_index = defaultdict(set)

# Read each text file
for filename in os.listdir(folder):

    if filename.endswith(".txt"):

        filepath = os.path.join(folder, filename)

        with open(filepath, "r", encoding="utf-8") as file:

            text = file.read().lower()

            # Remove punctuation
            text = re.sub(r'[^a-z0-9\s]', ' ', text)

            # Tokenization
            words = text.split()

            # Build inverted index
            for word in words:

                if word not in stop_words:
                    inverted_index[word].add(filename)

# Display the inverted index
print("INVERTED INDEX\n")

for term in sorted(inverted_index):
    print(f"{term} --> {sorted(inverted_index[term])}")

# Vocabulary Size
print("\nVocabulary Size:", len(inverted_index))
```

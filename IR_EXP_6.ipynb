```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import quote

def search_news(topic, websites):
    print("\nSearching for news related to:", topic)
    print("Selected websites:", ", ".join(websites))
    print("-" * 70)

    for website in websites:

        # Create Google News RSS search query
        search_query = quote(f"{topic} site:{website}")
        url = f"https://news.google.com/rss/search?q={search_query}&hl=en-IN&gl=IN&ceid=IN:en"

        try:
            response = requests.get(
                url,
                headers={"User-Agent": "Mozilla/5.0"},
                timeout=10
            )

            if response.status_code != 200:
                print(f"\nUnable to search {website}")
                continue

            soup = BeautifulSoup(response.content, "xml")

            print(f"\nNews from {website}")
            print("-" * 50)

            items = soup.find_all("item")

            if not items:
                print("No news stories found.")
                continue

            for i, item in enumerate(items[:5], 1):
                title = item.find("title")
                link = item.find("link")
                pub_date = item.find("pubDate")

                print(f"{i}. {title.text if title else 'No title'}")
                print(f"   Link: {link.text if link else 'No link'}")

                if pub_date:
                    print(f"   Date: {pub_date.text}")

                print()

        except requests.exceptions.RequestException as e:
            print(f"\nError accessing {website}: {e}")


# Main program

topic = input("Enter the news topic: ")

print("\nAvailable websites:")
print("1. bbc.com")
print("2. reuters.com")
print("3. thehindu.com")
print("4. indianexpress.com")
print("5. timesofindia.indiatimes.com")

choice = input(
    "\nEnter website numbers separated by commas "
    "(or enter 'all'): "
)

all_websites = [
    "bbc.com",
    "reuters.com",
    "thehindu.com",
    "indianexpress.com",
    "timesofindia.indiatimes.com"
]

if choice.lower() == "all":
    selected_websites = all_websites
else:
    selected_websites = []

    for number in choice.split(","):
        number = number.strip()

        if number.isdigit():
            index = int(number) - 1

            if 0 <= index < len(all_websites):
                selected_websites.append(all_websites[index])

if selected_websites:
    search_news(topic, selected_websites)
else:
    print("No valid websites selected.")
```

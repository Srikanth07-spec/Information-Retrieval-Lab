```python
import xml.etree.ElementTree as ET

# XML data
xml_data = """
<web>
    <page name="A" topic="IR">
        <link>B</link>
        <link>C</link>
    </page>
    <page name="B" topic="IR">
        <link>C</link>
        <link>D</link>
    </page>
    <page name="C" topic="AI">
        <link>A</link>
        <link>D</link>
    </page>
    <page name="D" topic="IR">
        <link>C</link>
    </page>
</web>
"""

# Parse XML
root = ET.fromstring(xml_data)

pages = {}
topics = {}

for page in root.findall("page"):
    name = page.get("name")
    topics[name] = page.get("topic")
    pages[name] = [link.text for link in page.findall("link")]

# Generate Web Graph
print("WEB GRAPH")
for page in pages:
    print(page, "->", pages[page])

# Select topic
topic = "IR"
names = list(pages.keys())
n = len(names)

# Initial PageRank
rank = {page: 1/n for page in names}

# Topic Specific PageRank
for _ in range(100):

    new_rank = {}

    for page in names:
        new_rank[page] = 0.15 / sum(
            1 for p in names if topics[p] == topic
        ) if topics[page] == topic else 0

    for page in names:
        if pages[page]:
            value = 0.85 * rank[page] / len(pages[page])

            for link in pages[page]:
                new_rank[link] += value

    rank = new_rank

# Display result
print("\nTOPIC SPECIFIC PAGERANK")
for page in sorted(rank, key=rank.get, reverse=True):
    print(page, ":", round(rank[page], 4))
```
